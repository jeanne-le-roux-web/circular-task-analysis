```python
# read the table written by MouseReMoCo at the end of the marker file (lines "RecXXX , ...")
table = []
with open(marker_file, "r") as f:
    for line in f:
        parts = [p.strip() for p in line.split(",")]
        if parts[0].startswith("Rec"):
            table.append([float(p) for p in parts[1:9]])
table = np.array(table[:n_records])   # keep only the records of the session

# our values, with the same columns
ours = np.column_stack([n_laps, Re, Te, error, MT, IDe, Be, IPe])

# largest difference (in absolute value) for each variable
names = ["nLaps", "Re", "Te", "error", "MT/lap", "IDe/lap", "Be", "IPe"]
diff = np.abs(ours - table)
for j in range(len(names)):
    print(f"{names[j]:8s}: max difference = {diff[:, j].max():.2f}")
```
